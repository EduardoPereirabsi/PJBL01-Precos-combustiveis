```python
import pandas as pd
```

Concatenação de arquivos CSV em um único DataFrame

```python
name_combustivel = ["diesel-gnv_2023", "glp_2023"]

for name in name_combustivel:
    
    df_combustivel = pd.DataFrame()

    for num in range(12):
        df = pd.read_csv(f"data/{name}-{num+1:02d}.csv", sep=";")
        df_combustivel = pd.concat([df_combustivel, df], axis=0)
    df_combustivel.to_csv(f"{name}.csv", sep=";", index=False)
```
